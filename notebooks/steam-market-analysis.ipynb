```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import kagglehub
import os
from IPython.display import display

# easy read 
pd.set_option('display.float_format', lambda x: '%.2f' % x)
# Download latest version
path = kagglehub.dataset_download("artermiloff/steam-games-dataset")
file_path = os.path.join(path, "games_march2025_full.csv")
```

```python
dataset = pd.read_csv(file_path)

# Display the dataset
display(dataset.head())

# Display basic information about the dataset
display(dataset.info())
```

##### Convert to optimally data type for each columns

```python
dataset['release_date'] = pd.to_datetime(dataset['release_date'])

# check
display(dataset['release_date'].head())
```

##### Check for missing values

```python
missing_data = dataset.isnull().sum()
missing_percent = (dataset.isnull().sum() / len(dataset)) * 100

missing_df = pd.DataFrame({
    'Missing Values': missing_data,
    'Percent (%)': missing_percent
})


missing_df = missing_df[missing_df['Missing Values'] > 0].sort_values(by='Missing Values', ascending=False)
missing_df = missing_df.reset_index().rename(columns={'index': 'Column Name'})

missing_df
```

##### Check for duplicate values

```python
duplicate_data = dataset.duplicated().sum()
duplicate_percent = (dataset.duplicated().sum() / len(dataset)) * 100

print(f"Duplicates rows: {duplicate_data}")
print(f"Duplizates rows as %: {duplicate_percent}")


# Deep check
duplicate_appids = dataset.duplicated(subset=['appid']).sum()
duplicate_names = dataset.duplicated(subset='name').sum()

print(f"Duplicates 'appid': {duplicate_appids}")
print(f"Duplicates 'name': {duplicate_names}")
```

##### Found that there 756 duplicates name its may be Remaster games or demo / beta games.
##### I decided to delete demo / beta games.

```python
to_remove = 'demo|beta'

#  orginal dataset is - dataset
dataset_filtered = dataset[~dataset['name'].str.contains(to_remove, case=False,na=False)]

print("before", len(dataset))
print("after", len(dataset_filtered))
```
